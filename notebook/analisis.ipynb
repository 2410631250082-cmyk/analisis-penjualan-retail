```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
```

```python
df["Tahun"] = df["Order Date"].dt.year
df["Order Date"] = df["Order Date"].dt.to_period("M")
```

```python
import matplotlib.pylot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
tahun = df.groupby("Tahun")["Sales"].sum().reset_index()

plt.figure(figsize=(8, 4))
plt.plot(tahun["Tahun"], tahun["Sales"], marker="o")
plt.xticks(tahun["Tahun"])
plt.title("Penjualan naik 50% dari 2015 ke 2018")
plt.ylabel("Total Sales")
plt.savefig("../images/tren_tahun.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(8, 4))
sns.barplot(data=region, x="Region", y="Sales", color="steelblue")
plt.title("West unggul, South paling rendah")
plt.savefig("../images/penjualan_region.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
top10 = (df.groupby("Sub-Category")["Sales"].sum()
           .sort_values(ascending=False).head(10).reset_index())

plt.figure(figsize=(8, 5))
sns.barplot(data=top10, y="Sub-Category", x="Sales", color="steelblue")
plt.title("Phones dan Chairs menyumbang sekitar 29% penjualan")
plt.savefig("../images/top10_subkategori.png", dpi=150, bbox_inches="tight")
plt.show()
```
